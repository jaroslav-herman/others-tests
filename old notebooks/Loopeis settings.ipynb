```python
import sys
from pathlib import Path
import os
from glob import glob

def replace_voltage(input_file, output_file, old_value="1.550", new_value="1.480"):
    input_path = Path(input_file)
    output_path = Path(output_file)

    # Read file content
    try:
        content = input_path.read_text(encoding="utf-8")
    except UnicodeDecodeError:
        # fallback for non-UTF8 EC-Lab files
        content = input_path.read_text(encoding="latin-1")

    # Check if value exists
    if old_value not in content:
        print(f"Warning: value {old_value} not found in file.")

    # Replace value
    new_content = content.replace(old_value, new_value)

    # Save to new file
    output_path.write_text(new_content, encoding="utf-8")

    print(f"Done. Saved modified file to: {output_file}")


# if __name__ == "__main__":
#     if len(sys.argv) != 4:
#         print("Usage: python replace_voltage.py input_file output_file new_value")
#         sys.exit(1)

#     input_file = sys.argv[1]
#     output_file = sys.argv[2]
#     new_value = sys.argv[3]

#     replace_voltage(input_file, output_file, old_value="1.550", new_value=new_value)
```
