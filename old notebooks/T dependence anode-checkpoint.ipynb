```python
def overpotential(I,R):
    u = []
    for index,i in enumerate(I):
        u.append(R[0]*I[0]+np.trapezoid(R[0:index+1],I[0:index+1]))
    u = np.array(u)/1000
    return u
```

```python
URc = []
for j in range(0,9):
    I = data[:,2][12*j:12*j+12]
    Rc = dataRc[2*j+1]

    urc = overpotential(I,Rc)
    URc.append(urc)
    # print(URo)
    plt.plot(I,urc,'-')
    plt.plot(I,I*Rc/1000,'x')


    
```
