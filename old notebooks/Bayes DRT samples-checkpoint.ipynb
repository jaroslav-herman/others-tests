```python
def get_header_lines(filepath):
    with open(filepath, "r", encoding="latin-1") as f:
        # Skip first line, read second
        f.readline()  
        line2 = f.readline().strip()
    # line2 looks like: "Nb header lines : 76"
    # split by ":" and take the right part
    try:
        n_header = int(line2.split(":")[1])
    except Exception:
        raise ValueError(f"Could not parse header line: {line2}")
    return n_header
```

```python
def get_EIS_setting(file_path):
    with open(file_path, "r") as file:
        E = 'Ewe'
        for line_number, line in enumerate(file, start=1):
            if "Potential control" in line:
                E = line.split(': ')[1][:-1]
            if "Current amplifier" in line:
                booster = line.split(': ')[1][:-1]
    return E,booster



def save_impedance_to_h5(file_in, file_out,cycles = None, cycle_col="cycle number", metadata=None):
    """
    df: pandas DataFrame with multiple spectra
    cycle_col: column that identifies each spectrum
    global_metadata: dict stored at file level (optional)
    """
    
    skip_rows = get_header_lines(file_in)
    with open(file_in, encoding='latin1') as f:
        df = pd.read_csv(f,skiprows = skip_rows-1,usecols=np.arange(0,30),delimiter = '\t')
    # print(df)
    E_control, booster = get_EIS_setting(file_in)
    if E_control == 'Ewe':
        columns = ['freq/Hz', 'Re(Z)/Ohm','-Im(Z)/Ohm','<Ewe>/V','time/s','<I>/mA']
    elif E_control == 'Ewe-Ece':
        columns = ['freq/Hz', 'Re(Zwe-ce)/Ohm','-Im(Zwe-ce)/Ohm','<Ewe>/V','time/s','<I>/mA']

    
    unique_vals = np.unique(df["cycle number"].values)
    if cycles is None:
        cycles = unique_vals
    with h5py.File(file_out, "a") as f:

        # Store global metadata
        if metadata is not None:
            for k, v in global_metadata.items():
                f.attrs[k] = v

        # Group by spectrum
        for cycle, group in df.groupby(cycle_col):
            if cycle in cycles:
                freq = (group['freq/Hz'].values)
                if len(freq) < 3:
                    continue
                
                group_name = f"spectrum_{(int(cycle)):02d}"
                # grp = f.create_group(f"spectrum_{(int(cycle)):02d}")
                if group_name in f:
                    grp = f[group_name]
                else:
                    grp = f.create_group(group_name)
                    
                grp.attrs['E_control'] = E_control
                grp.attrs['booster'] = booster
                grp.attrs['file'] = file_in

                subgroup_name = 'EIS data'
                if subgroup_name in grp:
                    pass
                else:
                    subgrp = grp.create_group(subgroup_name)
                    # Store each column as its own dataset
                    for col in group.columns:
                        # print(col)
                        # if col == cycle_col:
                            # continue  # already encoded in group name
                        if col in columns:
                            data = group[col].to_numpy()
                            col_split = col.split("/")
                            subgrp.create_dataset(str(col_split[0]+' ('+col_split[1]+')'), data=data)

                    # Optionally store per-spectrum metadata
                    subgrp.attrs["cycle number"] = cycle


def save_spectra_h5(filename, inv, cycle,series,sample_spectrum_number,metadata=None):
    """
    spectra: list of (energy_array, value_array)
    metadata_list: list of dicts (same length as spectra) or None
    """
    dec = np.floor(np.log10(metadata['max_cycle']))+1
 
    
    with h5py.File(filename, "a") as f:
        # for i, (tau, gamma) in enumerate(spectra):
            # f"{number}"
        
        group_name = "spectrum_"+str(int(cycle)).zfill(int(dec))
        # grp = f.create_group(f"spectrum_{(int(cycle)):02d}")
        if group_name in f:
            grp = f[group_name]
        else:
            grp = f.create_group(group_name)

        grp.attrs['series'] = series
        grp.attrs['sample_spectrum_number'] = sample_spectrum_number
        subgroup_name = 'DRT data'
        if subgroup_name in grp:
            subgrp = grp[subgroup_name]
        else:
            subgrp = grp.create_group(subgroup_name)
        

        
        if inv.fit_type == 'map':
            if 'MAP DRT' in subgrp:
                ssubgrp = subgrp['MAP DRT']
            else:
                ssubgrp = subgrp.create_group('MAP DRT')
            
            tau = inv.tau_plot
            gamma = inv.final_gamma   
            if 'MAP gamma' in ssubgrp:
                print(f'MAP DRT spectrum {sample_spectrum_number} already exists')
                ssubgrp['tau (s)'][:] = tau
                ssubgrp['MAP gamma'][:] = gamma
            else:
                print(f'Creating new MAP DRT spectrum no. {sample_spectrum_number}')
                ssubgrp.create_dataset("tau (s)", data = tau)
                ssubgrp.create_dataset("MAP gamma", data = gamma)
            if metadata is not None:
                for key, val in metadata.items():
                    ssubgrp.attrs[key] = val
                ssubgrp.attrs["cycle number"] = cycle
                
                
        elif inv.fit_type == 'bayes':
            if 'Bayes DRT' in subgrp:
                ssubgrp = subgrp['Bayes DRT']
            else:
                ssubgrp = subgrp.create_group('Bayes DRT')

            tau = inv.tau_plot
            gamma = inv.final_gamma   
            gamma_ci_low = inv.final_ci_lo
            gamma_ci_high = inv.final_ci_hi
            if 'Bayes gamma' in ssubgrp:
                print(f'BAYES DRT spectrum {sample_spectrum_number} already exists')
                ssubgrp['tau (s)'][:] = tau
                ssubgrp['Bayes gamma'][:] = gamma
                ssubgrp['Bayes gamma CI low'][:] = gamma_ci_low
                ssubgrp['Bayes gamma CI high'][:] = gamma_ci_high
            else:
                print(f'Creating new BAYES DRT spectrum no. {sample_spectrum_number}')
                ssubgrp.create_dataset("tau (s)", data = tau)
                ssubgrp.create_dataset("Bayes gamma", data = gamma)
                ssubgrp.create_dataset("Bayes gamma CI low", data = gamma_ci_low)   
                ssubgrp.create_dataset("Bayes gamma CI high", data = gamma_ci_high) 
                ssubgrp.attrs["percentile"] = inv.percentile
                
            if metadata is not None:
                for key, val in metadata.items():
                    ssubgrp.attrs[key] = val
                ssubgrp.attrs["cycle number"] = cycle


            

                    
#         if 'tau (s)' in subgrp:
#             print(f'DRT spectrum {sample_spectrum_number} already exists')
#             subgrp['tau (s)'][:] = tau
#             fit_type
#             subgrp['gamma map'][:] = gamma
#             # if metadata is not None:
#             #     for key, val in metadata.items():
#             #         subgrp.attrs[key] = val
            
#         else:
#             print(f'Creating new DRT spectrum no. {sample_spectrum_number}')
#             subgrp.create_dataset("tau (s)", data = tau)
#             subgrp.create_dataset("gamma", data = gamma)
#         if metadata is not None:
#             for key, val in metadata.items():
#                 subgrp.attrs[key] = val
#             subgrp.attrs["cycle number"] = cycle
```

```python
# series += 1
    # freq_vec_old = [1,2]
# file_in = "C:/Users/Herman/Desktop/V_Day3_procedure1_05_PEIS_C01.mpt"
# file_in = r"C:\Users\Herman\Desktop\MFF\Diplomka\Data z PEMWE\049_IV_IV_Ir etched star, Pt 7 nm + C 105 nm etched star\Data z PEMWE\IV_Day5_procedure1_05_PEIS.mpt"
def calculate_DRT_Bayes(file_in,file_out,mode = 'optimize', sample_spectrum_number=0,series = 0,cycles = None, lambda_value = 0):

    # from time import time
    
    series += 1
    freq_vec_old = [1,2]
    print(file_in)
    skip_rows = get_header_lines(file_in)
    with open(file_in, encoding='latin1') as f:
        # df = pd.read_csv(f,skiprows = skip_rows-1,usecols=np.arange(0,84),delimiter = '\t')
        df = pd.read_csv(f,skiprows = skip_rows-1,delimiter = '\t')
    unique_vals = np.unique(df["cycle number"].values)
    E = 0
    E_control, booster = get_EIS_setting(file_in)
    # print(E_control)

    
    if cycles is None:
        cycles = unique_vals
        # cycles = unique_vals[0::100]
    for i,val in enumerate(cycles):
        # print(tm.perf_counter())
        


        
        f = (df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), 'freq/Hz'].values)
        if len(f) < 3:
            continue
        start = tm.time()
        sample_spectrum_number += 1
        # Z_exp = (df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), 'Re(Z)/Ohm'].values - 1j*df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), '-Im(Z)/Ohm'].values)
        if E_control == 'Ewe':
            Z = (df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), 'Re(Z)/Ohm'].values - 1j*df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), '-Im(Z)/Ohm'].values)
            E_new = np.mean(df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), '<Ewe>/V'].values)
        elif E_control == 'Ewe-Ece':
            # Z = (df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), 'Re(Zwe-ce)/Ohm'].values - 1j*df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), '-Im(Zwe-ce)/Ohm'].values)
            Z = (df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), 'Re(Z)/Ohm'].values - 1j*df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), '-Im(Z)/Ohm'].values)
            
            E_new = np.mean(df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), '<Ewe-Ece>/V'].values)
            # Z_exp = np.flip(Z_exp)
            # freq_vec = np.flip(freq_vec)
        I = (df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), '<I>/mA'].values)
        time = (df.loc[(df["cycle number"] == val) & (df["freq/Hz"] != 0), 'time/s'].values)
        print(np.mean(I))
        
        if E- E_new> 0.2:
            series += 1
        E = E_new
        
        if np.diff(f).any() <  0:
            print('problem')
        else:
            pass

        if E < 1.51:
            f_min = 1
        elif 1.51<E<1.61:
            f_min = 3
        elif 1.61<E<1.71:
            f_min = 5
        else:
            f_min = 10

        f_max = 20000
        mask = (f > f_min)  & (f <= f_max)
        mask &= np.isfinite(Z.real) & np.isfinite(Z.imag)
        f = f[mask]
        Z = Z[mask]
        inv = Inverter()
        samples = 70
        warmup = 70
        inv.fit(f, Z, mode=mode,nonneg = True,outliers = True,init_from_ridge = True,samples = samples,warmup = warmup)
        # print(inv.final_gamma)
        metadata = {'f min (Hz)' : f_min,'f max (Hz)' : f_max ,'E (V)' : E,'I (mA)' : np.mean(I),'time (s)' : np.mean(time),
                      'Ro (Ohm)' : inv.R_inf,
                      'L (H)' : inv.inductance,'Nonneg' : True,
                      'Outliers' : True,'Mode' : mode, 'warmup' : warmup, 'samples' : samples, 
            'cycle' : int(val), 'sample_spectrum_number' : int(sample_spectrum_number), 'series': int(series), 'max_cycle': len(unique_vals)
           }
        
        save_spectra_h5(file_out, inv, val,series,sample_spectrum_number, metadata) 
        end = tm.time()
        print(f'Spectrum {i+1}/{len(cycles)} successfully saved in time {end - start} s')
    return int(sample_spectrum_number),int(series)
```
